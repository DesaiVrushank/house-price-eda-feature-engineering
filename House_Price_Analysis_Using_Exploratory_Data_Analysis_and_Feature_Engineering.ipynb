**Data Cleaning**

```python
df = df.drop_duplicates()

print(df.duplicated().sum())
```

```python
df.loc[df["area_sqft"] < 0, "area_sqft"] = np.nan

print(df[df["area_sqft"] < 0])
```

```python
df.loc[df["price_lakh"] < 0, "price_lakh"] = np.nan

print(df[df["price_lakh"] < 0])
```

```python
df["area_sqft"] = df["area_sqft"].fillna(df["area_sqft"].mean())
```

```python
df["bedrooms"] = df["bedrooms"].fillna(df["bedrooms"].mean())
```

```python
df["age_years"] = df["age_years"].fillna(df["age_years"].mean())
```

```python
df["distance_city_km"] = df["distance_city_km"].fillna(df["distance_city_km"].mean())
```

```python
df["price_lakh"] = df["price_lakh"].fillna(df["price_lakh"].mean())
```

```python
print(df.isnull().sum())
```

```python
# Feature Engineering
df["price_per_sqft"] = (df["price_lakh"] / df["area_sqft"])
df["price_per_sqft"]
```

```python
df["sqft_per_bedroom"] = (df["area_sqft"] / df["bedrooms"])
df["sqft_per_bedroom"]
```

```python
# Target = price_lakh
print(df["price_lakh"].value_counts())
```
